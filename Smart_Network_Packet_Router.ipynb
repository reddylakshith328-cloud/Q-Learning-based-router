```python
# ============================================================
# SMART NETWORK PACKET ROUTER USING Q-LEARNING
# ABILENE.GRAPHML DATASET
# ============================================================

# ============================================================
# 1. IMPORT LIBRARIES
# ============================================================

import random
import numpy as np
import networkx as nx
import matplotlib.pyplot as plt

from collections import defaultdict

SEED = 42

random.seed(SEED)
np.random.seed(SEED)


# ============================================================
# 2. DATASET PATH
# ============================================================

file_path = "/Users/lakshith/Downloads/Abilene.graphml"


# ============================================================
# 3. LOAD ABILENE GRAPHML TOPOLOGY
# ============================================================

G = nx.read_graphml(file_path)

print("===== ABILENE TOPOLOGY =====")
print("Number of nodes:", G.number_of_nodes())
print("Number of edges:", G.number_of_edges())

print("\nNodes:")
print(list(G.nodes()))

print("\nEdges:")
print(list(G.edges(data=True)))


# ============================================================
# 4. CONVERT GRAPH TO UNDIRECTED NETWORK
# ============================================================

if G.is_directed():
    G = G.to_undirected()

print("\nGraph directed:", G.is_directed())


# ============================================================
# 5. PREPROCESS GRAPHML ATTRIBUTES
# ============================================================

def numeric_value(value, default=1.0):
    """
    Convert GraphML attribute values to float.
    """
    try:
        return float(value)
    except:
        return default


def preprocess_graph(G):
    """
    Prepare GraphML topology for the Q-learning simulator.

    Each edge receives:
        capacity
        delay
        queue
        util
    """

    for u, v, data in G.edges(data=True):

        # ----------------------------------------------------
        # Capacity
        # ----------------------------------------------------
        capacity = None

        for key in [
            "capacity",
            "Capacity",
            "bandwidth",
            "Bandwidth",
            "bw",
            "BW"
        ]:
            if key in data:
                capacity = numeric_value(data[key], None)
                break

        if capacity is None:
            capacity = 10.0

        # ----------------------------------------------------
        # Delay
        # ----------------------------------------------------
        delay = None

        for key in [
            "delay",
            "Delay",
            "latency",
            "Latency"
        ]:
            if key in data:
                delay = numeric_value(data[key], None)
                break

        if delay is None:
            delay = 1.0

        # ----------------------------------------------------
        # Store simulation attributes
        # ----------------------------------------------------
        data["capacity"] = capacity
        data["delay"] = delay
        data["queue"] = 0.0
        data["util"] = 0.0

    return G


G = preprocess_graph(G)

print("\nGraph preprocessing completed.")


# ============================================================
# 6. DISPLAY EDGE ATTRIBUTES
# ============================================================

print("\n===== EDGE ATTRIBUTES =====")

for u, v, data in G.edges(data=True):
    print(
        f"{u} -- {v} | "
        f"Capacity={data['capacity']} | "
        f"Delay={data['delay']} | "
        f"Queue={data['queue']}"
    )


# ============================================================
# 7. DRAW ABILENE TOPOLOGY
# ============================================================

plt.figure(figsize=(12, 8))

pos = nx.spring_layout(G, seed=SEED)

nx.draw(
    G,
    pos,
    with_labels=True,
    node_size=700,
    font_size=9
)

plt.title("Abilene Network Topology")
plt.show()


# ============================================================
# 8. STATE DISCRETIZATION
# ============================================================

N_BINS = 4


def discretize(value, bins=N_BINS):
    """
    Normalize a continuous value into a discrete state bin.
    """

    value = max(0.0, min(1.0, float(value)))

    return min(
        int(value * bins),
        bins - 1
    )


# ============================================================
# 9. STATE ENCODING
# ============================================================

def encode_state(
    queue,
    delay,
    utilization,
    congestion
):
    """
    State:
        s = (q, d, u, c)

    q = queue length
    d = link delay
    u = bandwidth utilization
    c = congestion level
    """

    return (
        discretize(queue),
        discretize(delay),
        discretize(utilization),
        discretize(congestion)
    )


# ============================================================
# 10. GET CURRENT NETWORK STATE
# ============================================================

def get_state(G, node, destination):

    queues = []
    delays = []
    utilizations = []

    neighbors = list(G.neighbors(node))

    for neighbor in neighbors:

        data = G[node][neighbor]

        queues.append(
            float(data.get("queue", 0.0))
        )

        delays.append(
            float(data.get("delay", 1.0))
        )

        utilizations.append(
            float(data.get("util", 0.0))
        )

    # --------------------------------------------------------
    # Queue state
    # --------------------------------------------------------

    q = (
        np.mean(queues)
        if queues
        else 0.0
    )

    # --------------------------------------------------------
    # Delay state
    # --------------------------------------------------------

    d = (
        np.mean(delays)
        if delays
        else 0.0
    )

    # Normalize delay
    d = d / 10.0

    # --------------------------------------------------------
    # Utilization state
    # --------------------------------------------------------

    u = (
        np.mean(utilizations)
        if utilizations
        else 0.0
    )

    # --------------------------------------------------------
    # Congestion state
    # --------------------------------------------------------

    c = 0.0

    if queues:
        if max(queues) > 0.7:
            c = 1.0

    return encode_state(
        q,
        d,
        u,
        c
    )


# ============================================================
# 11. Q-LEARNING ROUTER
# ============================================================

class QLearningRouter:

    def __init__(
        self,
        alpha=0.1,
        gamma=0.9,
        epsilon=0.30,
        epsilon_min=0.01,
        epsilon_decay=0.98
    ):

        self.alpha = alpha
        self.gamma = gamma

        self.epsilon = epsilon
        self.epsilon_min = epsilon_min
        self.epsilon_decay = epsilon_decay

        self.q = defaultdict(float)


    # --------------------------------------------------------
    # ε-GREEDY ACTION SELECTION
    # --------------------------------------------------------

    def choose_action(
        self,
        state,
        actions
    ):

        if not actions:
            return None

        # Exploration
        if random.random() < self.epsilon:

            return random.choice(actions)

        # Exploitation
        values = [
            self.q[(state, action)]
            for action in actions
        ]

        best = max(values)

        best_actions = [
            action
            for action, value
            in zip(actions, values)
            if value == best
        ]

        return random.choice(best_actions)


    # --------------------------------------------------------
    # BELLMAN Q-LEARNING UPDATE
    # --------------------------------------------------------

    def update(
        self,
        state,
        action,
        reward,
        next_state,
        next_actions
    ):

        old = self.q[
            (state, action)
        ]

        future = max(
            (
                self.q[
                    (next_state, a)
                ]
                for a in next_actions
            ),
            default=0.0
        )

        target = (
            reward
            + self.gamma * future
        )

        self.q[
            (state, action)
        ] = (
            old
            + self.alpha
            * (target - old)
        )


    # --------------------------------------------------------
    # EPSILON DECAY
    # --------------------------------------------------------

    def decay_exploration(self):

        self.epsilon = max(
            self.epsilon_min,
            self.epsilon
            * self.epsilon_decay
        )


# ============================================================
# 12. REWARD FUNCTION
# ============================================================

def compute_reward(
    delay,
    loss,
    congestion,
    w_delay=1.0,
    w_loss=5.0,
    w_congestion=2.0
):

    return -(
        w_delay * delay
        + w_loss * loss
        + w_congestion * congestion
    )


# ============================================================
# 13. SHORTEST PATH NEXT-HOP
# ============================================================

def shortest_path_next_hop(
    G,
    src,
    dst,
    weight="delay"
):

    try:

        path = nx.shortest_path(
            G,
            src,
            dst,
            weight=weight
        )

        if len(path) > 1:

            return path[1]

        return src

    except nx.NetworkXNoPath:

        return None


# ============================================================
# 14. OSPF-LIKE NEXT HOP
# ============================================================

def ospf_next_hop(
    G,
    src,
    dst
):

    H = G.copy()

    # OSPF link cost = inverse capacity
    for u, v, data in H.edges(data=True):

        capacity = float(
            data.get(
                "capacity",
                1.0
            )
        )

        data["ospf_cost"] = (
            1.0 / max(
                capacity,
                1e-9
            )
        )

    try:

        path = nx.shortest_path(
            H,
            src,
            dst,
            weight="ospf_cost"
        )

        if len(path) > 1:

            return path[1]

        return src

    except nx.NetworkXNoPath:

        return None


# ============================================================
# 15. RIP-LIKE NEXT HOP
# ============================================================

def rip_next_hop(
    G,
    src,
    dst
):

    try:

        path = nx.shortest_path(
            G,
            src,
            dst
        )

        if len(path) > 1:

            return path[1]

        return src

    except nx.NetworkXNoPath:

        return None


# ============================================================
# 16. RESET NETWORK QUEUES
# ============================================================

def reset_network(G):

    for u, v, data in G.edges(data=True):

        data["queue"] = 0.0
        data["util"] = 0.0


# ============================================================
# 17. UPDATE NETWORK QUEUES
# ============================================================

def update_queues(G):

    for u, v, data in G.edges(data=True):

        data["queue"] = max(
            0.0,
            data.get("queue", 0.0)
            - 0.01
        )

        data["util"] = max(
            0.0,
            data.get("util", 0.0)
            - 0.01
        )


# ============================================================
# 18. INJECT CONGESTION
# ============================================================

def inject_congestion(
    G,
    probability=0.10
):

    for u, v, data in G.edges(data=True):

        if random.random() < probability:

            data["queue"] = min(
                1.0,
                data.get("queue", 0.0)
                + random.uniform(
                    0.2,
                    0.5
                )
            )

            data["util"] = min(
                1.0,
                data.get("util", 0.0)
                + random.uniform(
                    0.2,
                    0.5
                )
            )


# ============================================================
# 19. SIMULATE ONE PACKET
# ============================================================

def route_packet(
    G,
    agent,
    src,
    dst,
    train=True,
    policy="qlearning"
):

    current = src

    path = [src]

    path_delay = 0.0

    lost = False

    visited = {src}

    max_hops = (
        G.number_of_nodes()
        * 2
    )

    hops = 0


    # --------------------------------------------------------
    # Forward packet
    # --------------------------------------------------------

    while (
        current != dst
        and hops < max_hops
    ):

        # ----------------------------------------------------
        # State
        # ----------------------------------------------------

        state = get_state(
            G,
            current,
            dst
        )

        neighbors = [
            n
            for n in G.neighbors(current)
            if n not in visited
        ]


        if not neighbors:

            lost = True

            break


        # ----------------------------------------------------
        # Select next-hop
        # ----------------------------------------------------

        if policy == "qlearning":

            if train:

                action = agent.choose_action(
                    state,
                    neighbors
                )

            else:

                q_values = [
                    agent.q[
                        (state, n)
                    ]
                    for n in neighbors
                ]

                action = neighbors[
                    int(
                        np.argmax(
                            q_values
                        )
                    )
                ]


        elif policy == "ospf":

            action = ospf_next_hop(
                G,
                current,
                dst
            )

            if action not in neighbors:

                action = (
                    neighbors[0]
                    if neighbors
                    else None
                )


        elif policy == "rip":

            action = rip_next_hop(
                G,
                current,
                dst
            )

            if action not in neighbors:

                action = (
                    neighbors[0]
                    if neighbors
                    else None
                )

        else:

            action = None


        if action is None:

            lost = True

            break


        # ----------------------------------------------------
        # Link information
        # ----------------------------------------------------

        edge = G[
            current
        ][
            action
        ]


        # ----------------------------------------------------
        # Delay
        # ----------------------------------------------------

        base_delay = float(
            edge.get(
                "delay",
                1.0
            )
        )

        queue = float(
            edge.get(
                "queue",
                0.0
            )
        )

        link_delay = (
            base_delay
            * (1.0 + queue)
        )

        path_delay += link_delay


        # ----------------------------------------------------
        # Packet loss
        # ----------------------------------------------------

        if (
            queue > 0.85
            and random.random() < 0.15
        ):

            lost = True

            break


        # ----------------------------------------------------
        # Increase queue
        # ----------------------------------------------------

        edge["queue"] = min(
            1.0,
            queue + 0.05
        )


        # ----------------------------------------------------
        # Increase utilization
        # ----------------------------------------------------

        edge["util"] = min(
            1.0,
            edge.get(
                "util",
                0.0
            ) + 0.05
        )


        # ----------------------------------------------------
        # Congestion indicator
        # ----------------------------------------------------

        congestion = (
            1.0
            if edge["queue"] > 0.7
            else 0.0
        )


        # ----------------------------------------------------
        # Next state
        # ----------------------------------------------------

        next_state = get_state(
            G,
            action,
            dst
        )

        next_neighbors = [
            n
            for n in G.neighbors(action)
            if n not in visited
        ]


        # ----------------------------------------------------
        # Reward
        # ----------------------------------------------------

        reward = compute_reward(
            link_delay,
            1.0 if lost else 0.0,
            congestion
        )


        # ----------------------------------------------------
        # Q-table update
        # ----------------------------------------------------

        if (
            train
            and policy == "qlearning"
        ):

            agent.update(
                state,
                action,
                reward,
                next_state,
                next_neighbors
            )


        # ----------------------------------------------------
        # Move to next node
        # ----------------------------------------------------

        current = action

        path.append(current)

        visited.add(current)

        hops += 1


        # ----------------------------------------------------
        # Queue decay
        # ----------------------------------------------------

        update_queues(G)


    # ========================================================
    # DELIVERY RESULT
    # ========================================================

    delivered = (
        not lost
        and current == dst
    )

    return {
        "delivered": delivered,
        "lost": not delivered,
        "delay": path_delay,
        "hops": hops,
        "path": path
    }


# ============================================================
# 20. SIMULATE EPISODE
# ============================================================

def simulate_episode(
    G,
    agent,
    n_packets=150,
    train=True,
    policy="qlearning"
):

    nodes = list(G.nodes())

    total_delay = 0.0

    total_loss = 0

    delivered = 0

    delays = []

    hop_counts = []


    # --------------------------------------------------------
    # Generate packets
    # --------------------------------------------------------

    for _ in range(n_packets):

        src, dst = random.sample(
            nodes,
            2
        )


        result = route_packet(
            G,
            agent,
            src,
            dst,
            train=train,
            policy=policy
        )


        if result["delivered"]:

            delivered += 1

            total_delay += (
                result["delay"]
            )

            delays.append(
                result["delay"]
            )

            hop_counts.append(
                result["hops"]
            )

        else:

            total_loss += 1


    # --------------------------------------------------------
    # Metrics
    # --------------------------------------------------------

    avg_delay = (
        total_delay
        / max(
            delivered,
            1
        )
    )

    loss_rate = (
        total_loss
        / n_packets
        * 100
    )

    pdr = (
        delivered
        / n_packets
        * 100
    )


    # --------------------------------------------------------
    # Jitter
    # --------------------------------------------------------

    if len(delays) > 1:

        jitter = float(
            np.std(delays)
        )

    else:

        jitter = 0.0


    # --------------------------------------------------------
    # Throughput estimate
    # --------------------------------------------------------

    # Assumed packet size = 1500 bytes
    packet_size_bits = (
        1500 * 8
    )

    total_bits = (
        delivered
        * packet_size_bits
    )

    total_time_seconds = (
        max(
            total_delay,
            1e-6
        )
        / 1000.0
    )

    throughput_mbps = (
        total_bits
        / total_time_seconds
        / 1e6
    )


    return {
        "delay": avg_delay,
        "loss": loss_rate,
        "pdr": pdr,
        "jitter": jitter,
        "throughput": throughput_mbps,
        "delivered": delivered,
        "hops": (
            np.mean(hop_counts)
            if hop_counts
            else 0.0
        )
    }


# ============================================================
# 21. CREATE Q-LEARNING AGENT
# ============================================================

agent = QLearningRouter(
    alpha=0.1,
    gamma=0.9,
    epsilon=0.30,
    epsilon_min=0.01,
    epsilon_decay=0.98
)


# ============================================================
# 22. TRAIN Q-LEARNING ROUTER
# ============================================================

print(
    "=========================================="
)
print(
    "TRAINING Q-LEARNING ROUTER"
)
print(
    "=========================================="
)

history = []

episodes = 200


for episode in range(
    1,
    episodes + 1
):

    # Reset queues
    reset_network(G)

    # Add random congestion
    inject_congestion(
        G,
        probability=0.10
    )


    # Run episode
    result = simulate_episode(
        G,
        agent,
        n_packets=150,
        train=True,
        policy="qlearning"
    )


    # Decay epsilon
    agent.decay_exploration()


    # Store results
    history.append(
        (
            result["delay"],
            result["loss"],
            result["pdr"],
            result["jitter"],
            result["throughput"],
            agent.epsilon
        )
    )


    if episode % 20 == 0:

        print(
            f"Episode {episode:3d} | "
            f"Delay: {result['delay']:.2f} ms | "
            f"Loss: {result['loss']:.2f}% | "
            f"PDR: {result['pdr']:.2f}% | "
            f"Epsilon: {agent.epsilon:.4f}"
        )


# ============================================================
# 23. CONVERT TRAINING HISTORY
# ============================================================

history = np.array(
    history
)

training_delay = history[:, 0]
training_loss = history[:, 1]
training_pdr = history[:, 2]
training_jitter = history[:, 3]
training_throughput = history[:, 4]
epsilon_values = history[:, 5]


# ============================================================
# 24. Q-LEARNING CONVERGENCE: DELAY
# ============================================================

plt.figure(figsize=(10, 5))

plt.plot(
    range(
        1,
        episodes + 1
    ),
    training_delay
)

plt.xlabel(
    "Training Episode"
)

plt.ylabel(
    "Average Delay (ms)"
)

plt.title(
    "Q-Learning Training Convergence - Delay"
)

plt.grid()

plt.show()


# ============================================================
# 25. Q-LEARNING CONVERGENCE: PACKET LOSS
# ============================================================

plt.figure(figsize=(10, 5))

plt.plot(
    range(
        1,
        episodes + 1
    ),
    training_loss
)

plt.xlabel(
    "Training Episode"
)

plt.ylabel(
    "Packet Loss (%)"
)

plt.title(
    "Q-Learning Training Convergence - Packet Loss"
)

plt.grid()

plt.show()


# ============================================================
# 26. Q-LEARNING CONVERGENCE: PDR
# ============================================================

plt.figure(figsize=(10, 5))

plt.plot(
    range(
        1,
        episodes + 1
    ),
    training_pdr
)

plt.xlabel(
    "Training Episode"
)

plt.ylabel(
    "Packet Delivery Ratio (%)"
)

plt.title(
    "Q-Learning Training Convergence - PDR"
)

plt.grid()

plt.show()


# ============================================================
# 27. EPSILON DECAY
# ============================================================

plt.figure(figsize=(10, 5))

plt.plot(
    range(
        1,
        episodes + 1
    ),
    epsilon_values
)

plt.xlabel(
    "Training Episode"
)

plt.ylabel(
    "Epsilon"
)

plt.title(
    "Epsilon-Greedy Exploration Decay"
)

plt.grid()

plt.show()


# ============================================================
# 28. FINAL Q-LEARNING EVALUATION
# ============================================================

print(
    "\n=========================================="
)
print(
    "FINAL Q-LEARNING EVALUATION"
)
print(
    "=========================================="
)

reset_network(G)

qlearning_result = simulate_episode(
    G,
    agent,
    n_packets=500,
    train=False,
    policy="qlearning"
)

print(
    f"Average Delay: "
    f"{qlearning_result['delay']:.2f} ms"
)

print(
    f"Packet Loss: "
    f"{qlearning_result['loss']:.2f}%"
)

print(
    f"PDR: "
    f"{qlearning_result['pdr']:.2f}%"
)

print(
    f"Throughput: "
    f"{qlearning_result['throughput']:.2f} Mbps"
)

print(
    f"Jitter: "
    f"{qlearning_result['jitter']:.2f} ms"
)

print(
    f"Average Hops: "
    f"{qlearning_result['hops']:.2f}"
)


# ============================================================
# 29. OSPF EVALUATION
# ============================================================

print(
    "\n=========================================="
)

print(
    "OSPF EVALUATION"
)

print(
    "=========================================="
)

reset_network(G)

ospf_result = simulate_episode(
    G,
    agent,
    n_packets=500,
    train=False,
    policy="ospf"
)

print(
    f"Average Delay: "
    f"{ospf_result['delay']:.2f} ms"
)

print(
    f"Packet Loss: "
    f"{ospf_result['loss']:.2f}%"
)

print(
    f"PDR: "
    f"{ospf_result['pdr']:.2f}%"
)

print(
    f"Throughput: "
    f"{ospf_result['throughput']:.2f} Mbps"
)

print(
    f"Jitter: "
    f"{ospf_result['jitter']:.2f} ms"
)


# ============================================================
# 30. RIP EVALUATION
# ============================================================

print(
    "\n=========================================="
)

print(
    "RIP EVALUATION"
)

print(
    "=========================================="
)

reset_network(G)

rip_result = simulate_episode(
    G,
    agent,
    n_packets=500,
    train=False,
    policy="rip"
)

print(
    f"Average Delay: "
    f"{rip_result['delay']:.2f} ms"
)

print(
    f"Packet Loss: "
    f"{rip_result['loss']:.2f}%"
)

print(
    f"PDR: "
    f"{rip_result['pdr']:.2f}%"
)

print(
    f"Throughput: "
    f"{rip_result['throughput']:.2f} Mbps"
)

print(
    f"Jitter: "
    f"{rip_result['jitter']:.2f} ms"
)


# ============================================================
# 31. COMPARISON TABLE
# ============================================================

methods = [
    "OSPF",
    "RIP",
    "Q-Learning"
]

delay_values = [
    ospf_result["delay"],
    rip_result["delay"],
    qlearning_result["delay"]
]

loss_values = [
    ospf_result["loss"],
    rip_result["loss"],
    qlearning_result["loss"]
]

throughput_values = [
    ospf_result["throughput"],
    rip_result["throughput"],
    qlearning_result["throughput"]
]

pdr_values = [
    ospf_result["pdr"],
    rip_result["pdr"],
    qlearning_result["pdr"]
]

jitter_values = [
    ospf_result["jitter"],
    rip_result["jitter"],
    qlearning_result["jitter"]
]


import pandas as pd

results_df = pd.DataFrame({

    "Method": methods,

    "Average Delay (ms)": delay_values,

    "Packet Loss (%)": loss_values,

    "Throughput (Mbps)": throughput_values,

    "PDR (%)": pdr_values,

    "Jitter (ms)": jitter_values
})


print(
    "\n=========================================="
)

print(
    "PERFORMANCE COMPARISON"
)

print(
    "=========================================="
)

display(
    results_df
)


# ============================================================
# 32. DELAY COMPARISON
# ============================================================

plt.figure(figsize=(8, 5))

plt.bar(
    methods,
    delay_values
)

plt.xlabel(
    "Routing Method"
)

plt.ylabel(
    "Average Delay (ms)"
)

plt.title(
    "Average End-to-End Delay Comparison"
)

plt.grid(
    axis="y"
)

plt.show()


# ============================================================
# 33. PACKET LOSS COMPARISON
# ============================================================

plt.figure(figsize=(8, 5))

plt.bar(
    methods,
    loss_values
)

plt.xlabel(
    "Routing Method"
)

plt.ylabel(
    "Packet Loss (%)"
)

plt.title(
    "Packet Loss Comparison"
)

plt.grid(
    axis="y"
)

plt.show()


# ============================================================
# 34. THROUGHPUT COMPARISON
# ============================================================

plt.figure(figsize=(8, 5))

plt.bar(
    methods,
    throughput_values
)

plt.xlabel(
    "Routing Method"
)

plt.ylabel(
    "Throughput (Mbps)"
)

plt.title(
    "Throughput Comparison"
)

plt.grid(
    axis="y"
)

plt.show()


# ============================================================
# 35. PDR COMPARISON
# ============================================================

plt.figure(figsize=(8, 5))

plt.bar(
    methods,
    pdr_values
)

plt.xlabel(
    "Routing Method"
)

plt.ylabel(
    "Packet Delivery Ratio (%)"
)

plt.title(
    "Packet Delivery Ratio Comparison"
)

plt.grid(
    axis="y"
)

plt.show()


# ============================================================
# 36. JITTER COMPARISON
# ============================================================

plt.figure(figsize=(8, 5))

plt.bar(
    methods,
    jitter_values
)

plt.xlabel(
    "Routing Method"
)

plt.ylabel(
    "Jitter (ms)"
)

plt.title(
    "Jitter Comparison"
)

plt.grid(
    axis="y"
)

plt.show()


# ============================================================
# 37. IMPROVEMENT OVER OSPF
# ============================================================

delay_improvement = (
    (
        ospf_result["delay"]
        - qlearning_result["delay"]
    )
    / ospf_result["delay"]
    * 100
)

loss_improvement = (
    (
        ospf_result["loss"]
        - qlearning_result["loss"]
    )
    / max(
        ospf_result["loss"],
        1e-9
    )
    * 100
)

throughput_improvement = (
    (
        qlearning_result["throughput"]
        - ospf_result["throughput"]
    )
    / max(
        ospf_result["throughput"],
        1e-9
    )
    * 100
)

pdr_improvement = (
    qlearning_result["pdr"]
    - ospf_result["pdr"]
)

jitter_improvement = (
    (
        ospf_result["jitter"]
        - qlearning_result["jitter"]
    )
    / max(
        ospf_result["jitter"],
        1e-9
    )
    * 100
)


print(
    "\n=========================================="
)

print(
    "Q-LEARNING IMPROVEMENT OVER OSPF"
)

print(
    "=========================================="
)

print(
    f"Delay improvement: "
    f"{delay_improvement:.2f}%"
)

print(
    f"Packet-loss improvement: "
    f"{loss_improvement:.2f}%"
)

print(
    f"Throughput improvement: "
    f"{throughput_improvement:.2f}%"
)

print(
    f"PDR improvement: "
    f"{pdr_improvement:.2f} percentage points"
)

print(
    f"Jitter improvement: "
    f"{jitter_improvement:.2f}%"
)


# ============================================================
# 38. Q-TABLE INFORMATION
# ============================================================

print(
    "\n=========================================="
)

print(
    "Q-TABLE INFORMATION"
)

print(
    "=========================================="
)

print(
    "Number of learned state-action pairs:",
    len(agent.q)
)


# ============================================================
# 39. DISPLAY LEARNED Q-VALUES
# ============================================================

q_table_rows = []

for (
    state,
    action
), value in agent.q.items():

    q_table_rows.append({

        "State": state,

        "Action": action,

        "Q-value": value
    })


q_table_df = pd.DataFrame(
    q_table_rows
)

q_table_df = q_table_df.sort_values(
    by="Q-value",
    ascending=False
)

print(
    "\nTop learned Q-values:"
)

display(
    q_table_df.head(20)
)


# ============================================================
# 40. SAMPLE LEARNED ROUTING DECISIONS
# ============================================================

print(
    "\n=========================================="
)

print(
    "SAMPLE LEARNED NEXT-HOP DECISIONS"
)

print(
    "=========================================="
)

sample_nodes = list(
    G.nodes()
)[:5]

for node in sample_nodes:

    neighbors = list(
        G.neighbors(node)
    )

    if not neighbors:
        continue

    state = get_state(
        G,
        node,
        random.choice(
            list(G.nodes())
        )
    )

    values = {
        neighbor:
        agent.q[
            (state, neighbor)
        ]

        for neighbor in neighbors
    }

    print(
        f"\nNode: {node}"
    )

    print(
        "Q-values:",
        values
    )

    best_neighbor = max(
        values,
        key=values.get
    )

    print(
        "Selected next-hop:",
        best_neighbor
    )


# ============================================================
# 41. SAVE RESULTS
# ============================================================

results_df.to_csv(
    "abilene_routing_results.csv",
    index=False
)

q_table_df.to_csv(
    "abilene_q_table.csv",
    index=False
)

print(
    "\nResults saved:"
)

print(
    "abilene_routing_results.csv"
)

print(
    "abilene_q_table.csv"
)


# ============================================================
# 42. FINAL SUMMARY
# ============================================================

print(
    "\n=========================================="
)

print(
    "FINAL SUMMARY"
)

print(
    "=========================================="
)

print(
    "Dataset:",
    file_path
)

print(
    "Nodes:",
    G.number_of_nodes()
)

print(
    "Edges:",
    G.number_of_edges()
)

print(
    "Training episodes:",
    episodes
)

print(
    "Learning rate alpha:",
    agent.alpha
)

print(
    "Discount factor gamma:",
    agent.gamma
)

print(
    "Initial epsilon: 0.30"
)

print(
    "Final epsilon:",
    agent.epsilon
)

print(
    "Q-table size:",
    len(agent.q)
)

print(
    "\nTraining and evaluation completed."
)
```
